```python
# Manipulação de dados
import pandas as pd

# Requisições HTTP
import requests

# Processamento de HTML
from bs4 import BeautifulSoup
```

```python
# Página da empresa no Fundamentus
url = "https://www.fundamentus.com.br/detalhes.php?papel=BBAS3"

print(url)
```

```python
headers = {
    "User-Agent": (
        "Mozilla/5.0 (Windows NT 10.0; Win64; x64) "
        "AppleWebKit/537.36 (KHTML, like Gecko) "
        "Chrome/138.0.0.0 Safari/537.36"
    )
}

resposta = requests.get(url, headers=headers)

print(resposta.status_code)
```

```python
print(resposta.text[:1000])
```

```python
soup = BeautifulSoup(resposta.text, "html.parser")

print(type(soup))
```

```python
print(soup.title)
```

```python
# Encontrar todas as tabelas da página
tabelas = soup.find_all("table")

print(f"Foram encontradas {len(tabelas)} tabelas.")
```

```python
print(type(tabelas))
```

```python
print(type(tabelas[0]))
```
